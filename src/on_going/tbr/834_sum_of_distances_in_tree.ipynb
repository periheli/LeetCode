```python
from collections import deque
from typing import List
```

```python
class Solution:
    def sumOfDistancesInTree(self, n: int, edges: List[List[int]]) -> List[int]:
        if n == 1:
            return [0]

        if n == 2:
            return [1, 1]

        connection_dict = {edge: [] for edge in range(n)}
        for edge in edges:
            connection_dict[edge[0]] = connection_dict[edge[0]] + [edge[1]]
            connection_dict[edge[1]] = connection_dict[edge[1]] + [edge[0]]
        print(connection_dict)

        queue = deque([0])
        dp = [[]] * n
        dp[0] = [0] * n
        visited_vertex_set = set()
        while len(queue) > 0:
            cur_vertex = queue.popleft()
            new_vertexs = [
                vertex
                for vertex in connection_dict[cur_vertex]
                if vertex not in visited_vertex_set
            ]
            visited_vertex_set.add(cur_vertex)
            print(new_vertexs)
            for new_vertex in new_vertexs:
                dp[cur_vertex][new_vertex] = 1
                dp[new_vertex] = [0] * n
                for vertex, dist in enumerate(dp[cur_vertex]):
                    if vertex != new_vertex:
                        if dist > 0:
                            dp[new_vertex][vertex] = dist + 1
                            dp[vertex][new_vertex] = dist + 1
                    else:
                        dp[new_vertex][cur_vertex] = 1
                print(dp)

            queue.extend(new_vertexs)
            print(dp)

        return [sum(dist_list) for dist_list in dp]

    def sumOfDistancesInTree(self, n: int, edges: List[List[int]]) -> List[int]:
        tree = {edge: set() for edge in range(n)}
        res = [0] * n
        count = [1] * n
        for i, j in edges:
            tree[i].add(j)
            tree[j].add(i)

        def dfs(root, pre):
            for i in tree[root]:
                if i != pre:
                    dfs(i, root)
                    count[root] += count[i]
                    res[root] += res[i] + count[i]

        def dfs2(root, pre):
            for i in tree[root]:
                if i != pre:
                    res[i] = res[root] - count[i] + n - count[i]
                    dfs2(i, root)

        dfs(0, -1)
        dfs2(0, -1)
        return res
```
